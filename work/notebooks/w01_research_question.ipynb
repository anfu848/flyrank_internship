## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from pathlib import Path
import csv


def find_data_path():
    candidates = [
        Path("data/raw/content_refresh_anonymized.csv"),
        Path.cwd() / "data/raw/content_refresh_anonymized.csv",
        Path.cwd().parent / "data/raw/content_refresh_anonymized.csv",
        Path.cwd().parents[1] / "data/raw/content_refresh_anonymized.csv",
    ]
    for candidate in candidates:
        if candidate.exists():
            return candidate
    raise FileNotFoundError("Could not find the starter CSV file.")


path = find_data_path()
with path.open(newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

row_count = len(rows)
client_count = len({row["client_id"] for row in rows})
avg_ctr = round(sum(float(row["ctr"]) for row in rows) / row_count, 4)

print("rows:", row_count)
print("distinct clients:", client_count)
print("mean ctr:", avg_ctr)
```
